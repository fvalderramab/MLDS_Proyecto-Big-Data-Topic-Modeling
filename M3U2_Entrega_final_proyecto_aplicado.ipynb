## **1.2. Definición del diseño y/o arquitectura de la solución**
---

La solución pretende abordar el análisis de tweets desde la ingesta de datos hasta la visualización de resultados, optimizando tanto la interpretación como la escalabilidad.

### **1.2.1. Descripción**
---

**¿Cómo se estructura la solución propuesta?**

1.  Almacenamiento de Datos
    
    Los datos serán almacenados en una base de datos documental como MongoDB, aprovechando su flexibilidad para manejar datos semiestructurados como, en este caso, los tweets. Con esta herramienta se podrá, de ser necesario realizar consultas rápidas y específicas, como filtrar tweets por sentimiento o por palabras clave. Cada documento en la base de datos contendrá información clave como el texto del tweet, la fecha, la cantidad de "me gusta", la fuente del tweet y la etiqueta de sentimiento.
    
    Aunque los 22.360 registros son manejables, dado el contexto académico y las posibles limitaciones de los planes gratuitos en servicios como MongoDB Atlas, se evaluará la viabilidad de mantener la base de datos localmente dependiendo de las restricciones del hosting gratuito.

2. Preprocesamiento de los Datos

    Antes de realizar el modelado de tópicos, se limpiará y preparará el texto de los tweets. Esto incluye:
    - Eliminación de ruido: emojis, URLs, menciones, y caracteres especiales innecesarios.
    - Normalización: conversión del texto a minúsculas, eliminación de stopwords (palabras funcionales como "el", "de", etc.), y lematización (reducir palabras a su forma base).
    
    Estas tareas se implementarán usando bibliotecas como NLTK o SpaCy, que son estándares en el procesamiento de texto.

3. Agrupación por Sentimientos

    Con los datos limpios, se agruparán los tweets según las etiquetas de sentimiento (positivo, negativo, neutral). Esto permitirá aplicar el análisis de tópicos de manera segmentada para responder preguntas específicas de interés en cada categoría de sentimiento.

4. Modelado de Tópicos con LDA (Latent Dirichlet Allocation)

    Utilizando técnicas de aprendizaje no supervisado, se identificarán los temas predominantes dentro de cada grupo de sentimiento. Especificamente con LDA, el cual es un modelo probabilístico ampliamente usado en PLN. Con esto podremos asignar a cada tweet probabilidades de pertenencia a diferentes temas.
    
    Este modelo se implementará utilizando bibliotecas como Gensim o PySpark MLlib para aprovechar capacidades de procesamiento distribuido si se ve útil.

    Para asegurar interpretabilidad, se optimizará el número de temas mediante métodos como la coherencia de tópicos y pruebas iterativas.

5. Visualización Interactiva

    Los resultados del modelado se presentarán mediante gráficos interactivos que faciliten su interpretación haciendo uso de herramientas como Plotly que permitan integrar estas visualizaciones en una interfaz sencilla y accesible. Por ejemplo:

    - Distribución de tópicos por sentimiento.
    - Palabras clave asociadas a cada tema.
    - Relación entre los sentimientos y la frecuencia de ciertos temas.

**¿Cuáles son los componentes clave de la solución?**

1. Gestión de Datos:
MongoDB es una opción preferida para el almacenamiento debido a su capacidad de manejar datos semiestructurados como texto enriquecido y metadatos. Si las limitaciones del hosting gratuito resultan restrictivas, se evaluará usar una instancia local para garantizar la estabilidad y escalabilidad.

2. Procesamiento Distribuido:
Aunque el volumen de datos (22.360 filas) no parece ser un desafío significativo, la implementación de Dask puede ser útil para realizar análisis paralelos o en caso de una futura expansión de datos. Dask permite dividir tareas complejas, como el preprocesamiento o el modelado de tópicos, en múltiples nodos si fuera necesario.

3. Algoritmos de PLN:
LDA será el núcleo del análisis de tópicos. La elección de Gensim para la implementación permite trabajar eficientemente en sistemas locales.

4. Herramientas de Visualización:
Plotly, con su capacidad de generar gráficos dinámicos, es la herramienta clave para presentar los resultados. Estas visualizaciones garantizarán que los usuarios comprendan fácilmente los hallazgos.


**¿Existen opciones de diseño alternativas y cuáles son sus ventajas y desventajas?**

En caso de que el almacenamiento en la nube sea una limitante, se puede optar por PostgreSQL para manejar datos estructurados, trabajar localmente con MongoDB, o directamente tener archivos locales en formato CSV o Parquet. Sin embargo, estas alternativas pueden ser menos convenientes para consultas complejas o para crecimiento futuro.

#### **MongoDB Atlas sobre Cassandra, PostgreSQL y SQL**

**MongoDB Atlas** ha sido seleccionado como la base de datos para el almacenamiento y consulta de los tweets debido a su capacidad de manejar datos semiestructurados como JSON de manera natural. En un proyecto que analiza texto enriquecido con metadatos (como fecha, sentimiento y fuente), la flexibilidad de MongoDB supera las limitaciones estructurales de bases de datos relacionales como PostgreSQL o SQL.

Mientras que **Cassandra** es también una base de datos NoSQL robusta, su enfoque en almacenamiento de grandes volúmenes de datos distribuidos no se justifica en el caso de un dataset de 22.360 registros. Cassandra está diseñada para casos de uso donde la velocidad de escritura y la escalabilidad horizontal extrema son esenciales, características que exceden las necesidades actuales del proyecto.

Por otro lado, aunque **PostgreSQL** ofrece soporte para JSON, su estructura relacional puede complicar las consultas que involucran datos jerárquicos o dinámicos, como tweets con diferentes combinaciones de metadatos y texto. La simplicidad y flexibilidad de MongoDB son más adecuadas para iteraciones rápidas en un proyecto académico con énfasis en procesamiento de lenguaje natural.



#### Ventajas de MongoDB Atlas

- **Modelo documental**: MongoDB maneja documentos JSON-like de manera nativa, permitiendo almacenar cada tweet junto con sus metadatos sin requerir un esquema rígido.
- **Escalabilidad en la nube**: La implementación en Atlas facilita la gestión y escalabilidad, eliminando la necesidad de infraestructura local compleja.
- **Consultas eficientes**: Su soporte para consultas dinámicas y pipelines de agregación permite filtrar tweets por palabras clave, sentimiento o cualquier otro atributo.
- **Facilidad de integración**: MongoDB se integra fácilmente con Python a través de `pymongo`, simplificando la extracción y transformación de datos para tareas de preprocesamiento y análisis.



#### Limitaciones y su mitigación

Una posible limitación de MongoDB Atlas es el costo asociado con planes premium si los datos escalan significativamente. Sin embargo, los planes gratuitos actuales ofrecen suficiente capacidad para manejar este proyecto. En caso de superar estos límites, se podría optar por ejecutar una instancia local de MongoDB.



### **Dask sobre PySpark**

**Dask** ha sido seleccionado como la herramienta de procesamiento de datos debido a su integración nativa con el ecosistema Python y su capacidad para escalar desde entornos locales a distribuidos. Aunque **PySpark** es una opción robusta para análisis de Big Data, su configuración inicial y curva de aprendizaje son mayores, lo que lo hace menos adecuado para un proyecto académico con un tamaño de dataset manejable como este.

Con 22.360 tweets, Dask ofrece una solución ligera y eficiente para realizar tareas de preprocesamiento como tokenización, eliminación de ruido y lematización. Si bien PySpark sobresale en entornos distribuidos a gran escala, no se espera que el volumen de datos de este proyecto justifique su uso en esta etapa.



#### **Ventajas de Dask**

- **Familiaridad con Python**: Dask extiende pandas y NumPy, lo que facilita el uso al ya estar familiarizados con estas bibliotecas.
- **Ligero y escalable**: Permite el procesamiento paralelo en una máquina local y puede escalar a clústeres si se requiere.
- **Simplicidad en la instalación y uso**: A diferencia de PySpark, Dask no requiere la configuración de un entorno distribuido desde el inicio.
- **Flexibilidad académica**: Ideal para iteraciones rápidas en el desarrollo de técnicas de preprocesamiento y análisis.



#### Limitaciones y su mitigación

Dask no está optimizado para trabajar con volúmenes de datos masivos en clústeres grandes como PySpark. Sin embargo, esta limitación no afecta el proyecto. En caso de que los datos crezcan significativamente en futuros análisis, se podría migrar a PySpark para aprovechar su infraestructura distribuida.

### **Gensim vs Scikit learn?**

La elección de **Gensim** para el modelado de tópicos en lugar de **Scikit-learn** se fundamenta en su diseño especializado para tareas de procesamiento de lenguaje natural (PLN). Gensim permite manejar corpus grandes mediante streaming, evitando la carga completa en memoria, lo cual es importante para datasets que podrían escalar. Su integración nativa con herramientas de visualización como PyLDAvis lo convierte en una opción más robusta para proyectos de análisis de texto.

En contraste, Scikit-learn, aunque viable, está diseñado para tareas generalistas de machine learning. Su implementación de LDA requiere que los datos estén completamente en memoria, lo que limita su escalabilidad. Además, no cuenta con herramientas integradas para la representación de texto ni para tareas específicas de PLN, como tokenización, lo que añade pasos adicionales al flujo de trabajo.

Gensim no solo facilita el manejo de texto extenso, sino que también ofrece un enfoque modular y eficiente para tareas de análisis de tópicos, alineándose mejor con los objetivos del proyecto.


La elección de **MongoDB Atlas**, **Dask** y **Gensim** responde a los objetivos académicos y técnicos del proyecto, priorizando la flexibilidad, facilidad de uso y escalabilidad necesaria para analizar los datos existentes. MongoDB se alinea con la naturaleza semiestructurada de los tweets, mientras que Dask ofrece un entorno de procesamiento paralelo eficiente y fácil de implementar. Ambas tecnologías ofrecen un equilibrio óptimo entre rendimiento y simplicidad, asegurando que el proyecto cumpla sus metas dentro de las restricciones de tiempo y recursos disponibles.

### **Herramientas Adicionales**

**NLTK** y **PyLDAvis** complementan el análisis de tópicos al fortalecer el preprocesamiento y la visualización de resultados. NLTK es una biblioteca estándar en PLN que se utilizará para limpiar y preparar los tweets, eliminando ruido como URLs, emojis y stopwords, y tokenizando texto. Estas capacidades son esenciales para garantizar que el texto sea adecuado para el análisis de tópicos. Sin embargo el trabajo de lematización se va a delegar a la librería **spaCy**.

La razón de utilizar la librería **spaCy** en vez de NLTK para la lematización se debe principalmente al hecho de que spaCy está diseñado para manejar grandes volúmenes de texto de manera más eficiente, lo que lo hace ideal para proyectos que requieren análisis en tiempo real o en datasets grandes, como los 22,360 tweets del proyecto, en cambio NLTK, que tiene un enfoque más académico y modular. La mayor eficiencia de spaCy se debe a que está optimizado y preentrenado justamente para realizar tareas como la lematización y además utiliza modelos estadísticos avanzados que consideran el contexto gramatical de las palabras, proporcionando lemas más precisos en comparación con las reglas fijas de NLTK. Es decir que spaCy por ejemplo puede distinguir lemas cuando se usan como verbo y como sustantivo, mientras que NLTK a menudo depende de listas de palabras y reglas menos sofisticadas. También, spaCy incluye soporte nativo para varios idiomas y optimización mediante vectores de palabras y redes neuronales, lo que lo hace más adecuado para tareas complejas o en proyectos donde la precisión contextual es clave como el presente proyecto, en el que se pueden presentar una multitud de tópicos (debido a que el mundial es un tema tan globalizado) y es crítico tener un buen entendimiento de los diferentes contextos presentes en las opiniones recolectadas.

Finalmente, PyLDAvis facilita la interpretación de los modelos al proporcionar una visualización interactiva de los temas generados. Esto incluye la distribución de tópicos en el corpus y las palabras más representativas de cada uno, lo que permite explorar y comprender mejor los resultados del modelo. La combinación de estas herramientas asegura un flujo de trabajo completo, desde el preprocesamiento hasta la interpretación visual, maximizando la claridad y utilidad del análisis.

### 1. **Limpieza del ruido**:
Estas entidades suelen ser útiles para propósitos distintos al análisis semántico de tópicos y pueden introducir ruido en los modelos de texto. En especial, para el análisis de tópicos que queremos realizar. Por esta razón:

   - Convertimos el texto a minúsculas para evitar duplicación de palabras (por ejemplo, "Qatar" y "qatar") y facilitar la normalización del texto ya que hacer esto evita que el mismo término en diferentes casos sea interpretado como palabras distintas.

Para cada función de preprocesamiento se guardarán los datos en mongoDB. Se utiliza bulk_write para hacer el proceso rápido y eficiente

```python
def print_head(h=5):
    head = collection.find().limit(h)
    for doc in head:
        print(json.dumps(doc, indent=4, default=str, sort_keys=True))

def head(data, n=5):
    for tweet in data[:n]:
        print(json.dumps(tweet, indent=4, default=str, sort_keys=True))
```

```python
# Limpieza con regex
from pymongo import UpdateOne
def preprocess_tweet(tweet):

    tweet = re.sub(r'http\S+|www\S+|https\S+', '', tweet, flags=re.MULTILINE) # Eliminar enlaces
    tweet = re.sub(r'@\w+|#\w+', '', tweet) # Eliminar menciones y hashtags
    tweet = re.sub(r'[^A-Za-z\s]', ' ', tweet) # Eliminar símbolos, puntuación y números, manteniendo solo caracteres alfabéticos y espacios
    tweet = re.sub(r'\n', ' ', tweet) # Eliminar caracteres de nueva línea
    tweet = re.sub(r'\s+', ' ', tweet).strip() # Eliminar espacios múltiples
    tweet = tweet.lower() # Convertir a minúsculas
    return tweet

# Obtener todos los tweets de la colección
tweets_data = list(collection.find())

# Crear una lista de operaciones UpdateOne
operations = [
    UpdateOne({'_id': tweet['_id']}, {'$set': {'clean_text': preprocess_tweet(tweet['Tweet'])}})
    for tweet in tweets_data
]

# Actualizar todos los documentos en la colección usando bulk_write
collection.bulk_write(operations)
```

```python
from pymongo import UpdateOne
from nltk.tokenize import word_tokenize

def tokenize_tweet(tweet):
    return word_tokenize(tweet)

# Obtener todos los tweets de la colección
tweets_data = list(collection.find())

# Crear una lista de operaciones UpdateOne
operations = []
for tweet in tweets_data:
    if 'clean_text' in tweet:
        operations.append(
            UpdateOne({'_id': tweet['_id']}, {'$set': {'words': tokenize_tweet(tweet['clean_text'])}})
        )

# Actualizar todos los documentos en la colección usando bulk_write
collection.bulk_write(operations)
```

### 3. **Eliminación de Stopwords**:
   - Filtramos palabras comunes (como artículos u conectores) que no aportan valor semántico.
   

```python
from pymongo import UpdateOne
from gensim.utils import simple_preprocess
import nltk
from nltk.corpus import stopwords

nltk.download('stopwords')

stop_words = set(stopwords.words('english'))

def remove_stopwords(tweet):
    return [word for word in simple_preprocess(str(tweet)) if word not in stop_words]

# Obtener todos los tweets de la colección
tweets_data = list(collection.find())

# Crear una lista de operaciones UpdateOne
operations = []
for tweet in tweets_data:
    if 'words' in tweet:
        operations.append(
            UpdateOne({'_id': tweet['_id']}, {'$set': {'words': remove_stopwords(tweet['words'])}})
        )

# Actualizar todos los documentos en la colección usando bulk_write
collection.bulk_write(operations)
```

### 4. **Lematización**:
   - Reducir palabras a su forma base (por ejemplo, "jugadores" a "jugador") para agrupar variantes de la misma palabra. Esto ayuda a agrupar términos relacionados y mejora la detección de tópicos.

Usamos dask para trabajar de forma paralela y reducir el tiempo de la lematización

```python
!python -m pip install "dask[distributed]"
```

```python
import spacy
nlp = spacy.load("en_core_web_sm")
```

```python
import dask.bag as db
from dask.distributed import Client, progress
import pymongo

# 1. Adaptive Scheduling:
client = Client(processes=False, threads_per_worker=4,
                n_workers=4, memory_limit='auto')

def lemmatize_tweet(tweet, allowed_postags=['NOUN', 'ADJ', 'VERB', 'ADV']):
    if 'words' in tweet:
        doc = nlp(" ".join(tweet['words']))
        tweet['lemma'] = [token.lemma_ for token in doc if token.pos_ in allowed_postags]
    return tweet

# 2. Smaller Batches:
batch_size = 500

# 3. Error Handling and Progress Bar:
for i in range(0, len(tweets_data), batch_size):
    batch = tweets_data[i : i + batch_size]
    tweet_bag = db.from_sequence(batch)

    # Error Handling: Retry failed tasks
    lemmatized_batch = tweet_bag.map(lemmatize_tweet).compute(retries=3)

    operations = []  # Crea una lista de operaciones
    for tweet in lemmatized_batch:
         operations.append(
             pymongo.UpdateOne({'_id': tweet['_id']}, {'$set': {'lemma': tweet['lemma']}})
         )
    collection.bulk_write(operations)
```

### 5. **Creación de trigramas**:
   - Agrupar palabras en trigramas puede ayudar a eliminar ruido y redundancia, haciendo que los análisis posteriores sean más precisos.

```python
import gensim
tweets_data = list(collection.find())
tokenized_tweets = [tweet['lemma'] for tweet in tweets_data if 'lemma' in tweet]

# Crear bigramas
bigram = gensim.models.Phrases(tokenized_tweets, min_count=5, threshold=100)
bigram_mod = gensim.models.phrases.Phraser(bigram)

# Crear trigramas
trigram = gensim.models.Phrases(bigram[tokenized_tweets], threshold=100)
trigram_mod = gensim.models.phrases.Phraser(trigram)

# Crear una lista de operaciones UpdateOne
operations = []
for tweet in tweets_data:
    if 'lemma' in tweet:
        tweet['trigrams'] = trigram_mod[bigram_mod[tweet['lemma']]]  # Aplicar trigramas
        operations.append(
            UpdateOne({'_id': tweet['_id']}, {'$set': {'trigrams': tweet['trigrams']}})
        )

# Actualizar todos los documentos en la colección usando bulk_write
collection.bulk_write(operations)

# Extraer todas las combinaciones de trigramas únicos (después de la actualización)
all_trigrams = set()
for tweet in tweets_data:
    if 'trigrams' in tweet:
        for trigram in tweet['trigrams']:
            if '_' in trigram:
                all_trigrams.add(trigram)

print("Trigramas únicos:", list(all_trigrams))
```

A continuación, preparamos el corpus y el diccionario necesarios para entrenar el modelo LDA utilizando la biblioteca `gensim`. Primero, extraemos los textos tokenizados desde la columna `trigrams` del DataFrame, que contiene las palabras o frases preprocesadas del conjunto de datos. Luego, creamos un diccionario con `corpora.Dictionary`, que asigna un identificador único a cada palabra en el corpus. Para reducir el ruido en los datos, se utiliza `filter_extremes` con los parámetros `no_below=5`, para excluir palabras que aparecen en menos de 5 documentos, y `no_above=0.5`, para eliminar términos que están presentes en más del 50% de los documentos, generalmente palabras muy comunes y menos informativas. Finalmente, convertimos los textos en una representación de bolsa de palabras (Bag-of-Words) mediante `doc2bow`, que transforma cada documento en una lista de pares `(word_id, word_count)`, preparándolo para el entrenamiento del modelo LDA.

```python
import gensim
from gensim import corpora
texts = df_cleaned['trigrams'].tolist()
dictionary = corpora.Dictionary(texts)
dictionary.filter_extremes(no_below=5, no_above=0.5)
corpus = [dictionary.doc2bow(text) for text in texts]
```

Ahora, definimos la función `compute_coherence_values`, diseñada para calcular los puntajes de coherencia para los modelos LDA entrenados con diferentes números de tópicos. La función toma como parámetros el diccionario, el corpus, los textos preprocesados, y una serie de valores de control como el rango de tópicos `(start, limit, step)`, así como configuraciones del modelo (chunksize y passes). Iteramos sobre la cantidad de tópicos en el rango especificado, entrenando un modelo LDA para cada valor de `num_topics`. Evaluamos cada modelo utilizando la métrica de coherencia `c_v`, que mide qué tan interpretables son los temas generados.

```python
# Function to compute coherence score
from gensim.models import CoherenceModel
def compute_coherence_values(dictionary, corpus, texts, limit, start=2, step=3, chunksize=1000, passes=20):
    coherence_values = []
    model_list = []
    for num_topics in range(start, limit, step):
        model = gensim.models.ldamodel.LdaModel(corpus=corpus,
                                                id2word=dictionary,
                                                num_topics=num_topics, # Hay que variar número de temas (cambiando los valores de start, limit y step) (entre más, más tiempo de entrenamiento)
                                                random_state=100,
                                                update_every=1, # Si no está convergiendo bien, se puede disminuir para actualizar el modelo más frecuentemente
                                                chunksize=chunksize, # Entre mayor, acelera la convergencia (más memoria). Menor, más cantidad de iteraciones para converger
                                                passes=passes, # Entre mayor, el modelo puede converger mejor (más tiempo de entrenamiento)
                                                alpha='auto', # Si tweets están asociados a muchos temas, disminuirlo. Si no, se puede aumentar
                                                per_word_topics=True
                                                #, eta=eta # Si los temas contienen muchas palabras irrelevantes, se aumenta para que sean más específicos.
                                                )
        model_list.append(model)
        coherencemodel = CoherenceModel(model=model, texts=texts, dictionary=dictionary, coherence='c_v')
        coherence_values.append(coherencemodel.get_coherence())
    return model_list, coherence_values

# Compute coherence values for different number of topics
model_list, coherence_values = compute_coherence_values(dictionary=dictionary, corpus=corpus, texts=texts, start=3, limit=13, step=1, chunksize=2000, passes=50)
```

De manera análoga a la anterior sección, preparamos el corpus y el diccionario necesarios para entrenar el modelo LDA mediante `gensim`; y calculamos los valores de coherencia.

```python
import gensim
from gensim import corpora
texts_positive = [dictionary['trigrams'] for dictionary in positive]
dictionary_positive = corpora.Dictionary(texts_positive)
dictionary_positive.filter_extremes(no_below=5, no_above=0.5)
corpus_positive = [dictionary_positive.doc2bow(text) for text in texts_positive]
```

```python
# Function to compute coherence score
from gensim.models import CoherenceModel
def compute_coherence_values(dictionary, corpus, texts, limit, start=2, step=3, chunksize=1000, passes=20):
    coherence_values = []
    model_list = []
    for num_topics in range(start, limit, step):
        model = gensim.models.ldamodel.LdaModel(corpus=corpus,
                                                id2word=dictionary,
                                                num_topics=num_topics, # Hay que variar número de temas (cambiando los valores de start, limit y step) (entre más, más tiempo de entrenamiento)
                                                random_state=100,
                                                update_every=1, # Si no está convergiendo bien, se puede disminuir para actualizar el modelo más frecuentemente
                                                chunksize=chunksize, # Entre mayor, acelera la convergencia (más memoria). Menor, más cantidad de iteraciones para converger
                                                passes=passes, # Entre mayor, el modelo puede converger mejor (más tiempo de entrenamiento)
                                                alpha='auto', # Si tweets están asociados a muchos temas, disminuirlo. Si no, se puede aumentar
                                                per_word_topics=True
                                                #, eta=eta # Si los temas contienen muchas palabras irrelevantes, se aumenta para que sean más específicos.
                                                )
        model_list.append(model)
        coherencemodel = CoherenceModel(model=model, texts=texts, dictionary=dictionary, coherence='c_v')
        coherence_values.append(coherencemodel.get_coherence())
    return model_list, coherence_values

# Compute coherence values for different number of topics
model_list_positive, coherence_values_positive = compute_coherence_values(dictionary=dictionary_positive, corpus=corpus_positive, texts=texts_positive, start=2, limit=13, step=1, chunksize=2000, passes=50)
```

```python
import gensim
from gensim import corpora
texts_negative = [dictionary['trigrams'] for dictionary in negative]
dictionary_negative = corpora.Dictionary(texts_negative)
dictionary_negative.filter_extremes(no_below=5, no_above=0.5)
corpus_negative = [dictionary_negative.doc2bow(text) for text in texts_negative]
```

```python
# Function to compute coherence score
from gensim.models import CoherenceModel
def compute_coherence_values(dictionary, corpus, texts, limit, start=2, step=3, chunksize=1000, passes=20):
    coherence_values = []
    model_list = []
    for num_topics in range(start, limit, step):
        model = gensim.models.ldamodel.LdaModel(corpus=corpus,
                                                id2word=dictionary,
                                                num_topics=num_topics, # Hay que variar número de temas (cambiando los valores de start, limit y step) (entre más, más tiempo de entrenamiento)
                                                random_state=100,
                                                update_every=1, # Si no está convergiendo bien, se puede disminuir para actualizar el modelo más frecuentemente
                                                chunksize=chunksize, # Entre mayor, acelera la convergencia (más memoria). Menor, más cantidad de iteraciones para converger
                                                passes=passes, # Entre mayor, el modelo puede converger mejor (más tiempo de entrenamiento)
                                                alpha='auto', # Si tweets están asociados a muchos temas, disminuirlo. Si no, se puede aumentar
                                                per_word_topics=True
                                                #, eta=eta # Si los temas contienen muchas palabras irrelevantes, se aumenta para que sean más específicos.
                                                )
        model_list.append(model)
        coherencemodel = CoherenceModel(model=model, texts=texts, dictionary=dictionary, coherence='c_v')
        coherence_values.append(coherencemodel.get_coherence())
    return model_list, coherence_values

# Compute coherence values for different number of topics
model_list_negative, coherence_values_negative = compute_coherence_values(dictionary=dictionary_negative, corpus=corpus_negative, texts=texts_negative, start=2, limit=13, step=1, chunksize=2000, passes=50)
```
